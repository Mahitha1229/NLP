```python
import random
import os

def load_sentences(file_path):
    # Read all non-empty lines from the input file
    with open(file_path, 'r', encoding='utf-8') as f:
        return [line.strip() for line in f if line.strip()]

def save_sentences(sentences, file_path):
    # Write each sentence to a new line in the output file
    with open(file_path, 'w', encoding='utf-8') as f:
        for s in sentences:
            f.write(s + '\n')

def split_data(file_path):
    # Create output folder if it doesn't exist
    os.makedirs('data_splits', exist_ok=True)

    # Load and shuffle the data
    data = load_sentences(file_path)
    random.seed(42)
    random.shuffle(data)

    # Split into test, validation, and training sets
    test = data[:500]
    val = data[500:1000]
    train = data[1000:]

    # Save each split to its respective file
    save_sentences(train, 'data_splits/train.txt')
    save_sentences(val, 'data_splits/val.txt')
    save_sentences(test, 'data_splits/test.txt')

# Run the split function on your dataset
split_data('tokenized_telugu.txt')
```
